```python
import sys
import os

sys.path.append(os.path.abspath(".."))
import pandas as pd
from preprocessing.validator import validate_dataset, print_sufficiency_report, DatasetInsufficientError

# Test 1: a dataset missing Cost and Region (soft-required)
sample_ok = pd.DataFrame({
    "Date": ["2026-06-24", "2026-07-01", "2026-07-08", "2026-07-15"],
    "Store": ["Store A"] * 4,
    "Category": ["Electronics"] * 4,
    "Product": ["Laptop"] * 4,
    "Revenue": [50000, 48000, 47000, 30000],
    "Quantity": [5, 5, 4, 3],
})

try:
    report = validate_dataset(sample_ok)
    print_sufficiency_report(report)
except DatasetInsufficientError as e:
    print(e)

# Test 2: a dataset missing a hard-required column (Revenue)
sample_bad = sample_ok.drop(columns=["Revenue"])

try:
    report = validate_dataset(sample_bad)
    print_sufficiency_report(report)
except DatasetInsufficientError as e:
    print(e)
```
